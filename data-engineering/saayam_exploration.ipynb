```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load the dataset
df = pd.read_csv('team_hours.csv')
print('Dataset loaded successfully!')
```

```python
# 2. Show first 5 rows
print('First 5 rows:')
df.head()
```

```python
# 3. Count number of contributors
contributor_count = df['Name'].nunique()
print(f'Contributor count: {contributor_count}')
```

```python
# 4. Total hours worked
total_hours = df['Hours_Worked'].sum()
print(f'Total hours worked: {total_hours}')
```

```python
# 5. Group by Team — total hours per team + team with most hours
team_hours = df.groupby('Team')['Hours_Worked'].sum().reset_index()
team_hours.columns = ['Team', 'Total Hours']
print('Total hours per team:')
print(team_hours.to_string(index=False))

top_team = team_hours.loc[team_hours['Total Hours'].idxmax()]
print(f"\nTeam with most hours: {top_team['Team']} ({top_team['Total Hours']} hours)")
```

```python
# 6. Sort contributors by hours and show top 3
top_contributors = df.groupby('Name')['Hours_Worked'].sum().reset_index()
top_contributors = top_contributors.sort_values('Hours_Worked', ascending=False).head(3)
print('Top 3 contributors:')
for _, row in top_contributors.iterrows():
    print(f"  {row['Name']} ({row['Hours_Worked']} hours)")
```

```python
# 7. Bar chart of total hours per team
plt.figure(figsize=(10, 6))
plt.bar(team_hours['Team'], team_hours['Total Hours'], color='steelblue', edgecolor='black')
plt.title('Total Hours Worked per Team', fontsize=14)
plt.xlabel('Team', fontsize=12)
plt.ylabel('Total Hours', fontsize=12)
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```
